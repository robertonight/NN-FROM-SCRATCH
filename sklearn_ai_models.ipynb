## Import des librairies necéssaires à ce projet


```python
import numpy as np
```

## Import du dataset sérialisé

Les données de l'ensemble B sont récupérés à l'aide de la librairie *pickle*

```python
import pickle

with open("X.pickle", "rb") as f:
    X: np.ndarray = np.array(pickle.load(f))
    """
    ndarray 2D [height * width,...] contenant les données d'entrée
    """
with open("y.pickle", "rb") as f:
    y: np.ndarray = np.array(pickle.load(f))
    """
    ndarray 1D de classes identifiant de chaque donnée d'entrée "x"
    """
with open("filenames.pickle", "rb") as f:
    filenames: list = pickle.load(f)
    """
    liste des noms de chaque fichier de donnée (ex.: '301Triangle3.png')
    """
```

```python
RATIO_DATASET = 1
NB_OF_IMGS = int(len(X) * RATIO_DATASET)
"""nombre d'imgs pour faire l'entrainement"""


# convertir les vecteurs en type numpy
X = X[:NB_OF_IMGS]
X = np.array(X)
print("images d'entrée:", X.shape)

y = y[:NB_OF_IMGS]
y = np.array(y)
print("étiquettes d'entrée:", y.shape)

# X = X.reshape(-1, 40, 40)
```

### Encodage des des classes cibles
Dans cette cellule, les identifiants des classes en chaînes de caratères sont encodés sur des entiers.


```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()
y_encoded: np.ndarray = label_encoder.fit_transform(y)
"""
classes str de tout le dataset encodées sur des entiers
"""

print(f"Exemple :")
for i in range(3):
    print(f"\ty[{i}] = {y[i]}  \t -> y_encoded[{i}] = {y_encoded[i]}")
```

### Méthode Hold out (80% training, 20% tests)

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,  #
    test_size=0.2,  # 20% dedicated au test
    random_state=42,
    # Utilisation de 42 par convention pour garder une ref des données dans leurs split respectif en toutt temps
    stratify=y_encoded  # garde les memes proportions des classes selon le dataset original
)

X_train: np.ndarray
"""
données d'entrainement pour bâtir les models
"""
y_train: np.ndarray
"""
classes cibles (encodées sur des entiers) pour les données d'entrainement
"""

X_test: np.ndarray
"""
données de test pour determiner la performance des models
"""
y_test: np.ndarray
"""
classes cibles (encodées sur des entiers) pour les données des test
"""
```

```python
from sklearn.preprocessing import StandardScaler

# normaliser les données
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

### Metriques d'évaluation de models

Chaque model d'intelligence artificielle a besoin du paquet **metrics** de scikit-learn pour mesurer les prédictions données

```python
from sklearn import metrics
```

---


## Les k voisins le plus proches (k-NN)


```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn3 = KNeighborsClassifier(n_neighbors=3)
knn3.fit(X_train_scaled, y_train)
```

```python
knn5 = KNeighborsClassifier(n_neighbors=5)
knn5.fit(X_train_scaled, y_train)
```

```python
knn7 = KNeighborsClassifier(n_neighbors=7)
knn7.fit(X_train_scaled, y_train)
```

```python
y_prediction3 = knn3.predict(X_test_scaled)
real_accuracy3 = metrics.accuracy_score(y_test, y_prediction3)
print(f"knn avec 3 voisins: {real_accuracy3 * 100}%")
```

```python
y_prediction5 = knn5.predict(X_test_scaled)
real_accuracy5 = metrics.accuracy_score(y_test, y_prediction5)
print(f"knn avec 5 voisins: {real_accuracy5 * 100}%")
```

```python
y_prediction7 = knn7.predict(X_test_scaled)
real_accuracy7 = metrics.accuracy_score(y_test, y_prediction7)
print(f"knn avec 7 voisins: {real_accuracy7 * 100}%")
```

## Machine à vecteurs de support (SVM)

```python
from sklearn.svm import SVC
```

```python
# Radial Basis function (rbf) a été choisie à cause notre problème n'est pas linéaire (classification)
# https://medium.com/@suvigya2001/the-gaussian-rbf-kernel-in-non-linear-svm-2fb1c822aae0
svm = SVC(kernel="rbf", C=1.0)
svm.fit(X_train_scaled, y_train)
```

```python
y_prediction = svm.predict(X_test_scaled)
print(f"Accuracy: {metrics.accuracy_score(y_test, y_prediction) * 100}%")
```

```python
print(f"Classification report:")
print(metrics.classification_report(y_test, y_prediction, target_names=label_encoder.classes_))
```

```python
print(f"confusion matrix:")
print(metrics.confusion_matrix(y_test, y_prediction))
```

## Réseaux de neurones de SciKit Learn

```python
from sklearn.neural_network import MLPClassifier
```

```python
mlp = MLPClassifier(hidden_layer_sizes=(128,), activation='relu', solver='adam', max_iter=300, random_state=42)
```

```python
mlp.fit(X_train_scaled, y_train)
```

```python
y_prediction = mlp.predict(X_test_scaled)
```

```python
print(f"Accuracy: {metrics.accuracy_score(y_test, y_prediction) * 100}%")
```

```python
print(f"Classification report:")
print(metrics.classification_report(y_test, y_prediction, target_names=label_encoder.classes_))
```

```python
print(f"confusion matrix:")
print(metrics.confusion_matrix(y_test, y_prediction))
```
